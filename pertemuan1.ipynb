## 1. Contoh Alur Scraping dengan Selenium

Pola dasar:

1. Membuka browser;
2. Membuka URL;
3. Mencari elemen HTML;
4. Mengambil teks;
5. Memasukkan hasil ke DataFrame.

Selector seperti `class name` harus disesuaikan dengan struktur HTML website saat digunakan.

````python
# Contoh dasar penggunaan Selenium
# Jalankan jika Selenium dan Chrome sudah siap

import time
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By

options = Options()
options.add_argument("--disable-gpu")

driver = webdriver.Chrome(options=options)

# Membuka halaman website
driver.get("https://www.cnnindonesia.com/")

time.sleep(2)

print("Judul halaman:", driver.title)

# Variabel 'articles' harus didefinisikan dulu sebelum di-looping
articles = driver.find_elements(By.TAG_NAME, "h2")
titles = [article.text for article in articles if article.text.strip() != ""]

driver.quit()

# Simpan ke file CSV
df = pd.DataFrame(titles, columns=['judul'])
df.to_csv('scrapping_judul_berita_cnn.csv', index=False)

print(f"Sukses simpan {len(df)} judul berita ke CSV!")
display(df.head())
````

# MODUL 2 — Data Cleaning & Preprocessing

Data mentah jarang berada dalam kondisi sempurna dan siap digunakan.

Kita dapat menemukan:
- nilai yang hilang (**missing value**);
- data yang berulang (**duplicate**);
- tipe data yang kurang sesuai/anomali;
- perbedaan skala antar fitur.

**Data Cleaning** berfokus pada menemukan dan memperbaiki masalah data.

**Data Preprocessing** mengubah data yang sudah bersih menjadi bentuk yang lebih sesuai untuk analisis atau machine learning.

## 1. Missing Value

Missing value adalah data yang tidak tersedia atau tidak tercatat.

Pandas dapat merepresentasikannya sebagai `NaN`, `None`, atau `NaT`.

Ada dua pendekatan utama:
1. **Deletion** → menghapus baris/kolom yang bermasalah.
2. **Imputation** → mengisi nilai yang hilang.

Pilihan metode harus disesuaikan dengan kondisi data. Jangan asal menghapus data karena informasi yang masih berguna bisa ikut hilang.

````python
# Melihat jumlah missing value pada setiap kolom
missing = df.isnull().sum()

print(missing[missing > 0])
````

Kasus pada Dataset Titanic, kolom `deck` memiliki banyak sekali nilai kosong. Untuk contoh ini, kita hapus saja kolom ini.

Untuk `age`, kita gunakan **median** saja karena median lebih tahan terhadap pengaruh nilai ekstrem.

Untuk data kategorikal seperti `embarked`, kita gunakan **modus** saja.

````python
# Buat salinan agar dataset ori tetap tersimpan
df_clean = df.copy()

# Hapus kolom 'deck' karena missing value-nya sangat banyak
df_clean = df_clean.drop(columns=["deck"])

# Isi missing value pada 'age' dengan median
df_clean["age"] = df_clean["age"].fillna(df_clean["age"].median())

# Isi data 'embarked' dengan modus
df_clean["embarked"] = df_clean["embarked"].fillna(
    df_clean["embarked"].mode()[0]
)

df_clean["embark_town"] = df_clean["embark_town"].fillna(
    df_clean["embark_town"].mode()[0]
)

# Recheck missing value pada dataset
print(df_clean.isnull().sum()[df_clean.isnull().sum() > 0])
````

### Alternatif: Menghapus Baris yang Missing

Jika jumlah missing value sangat kecil dan baris tersebut tidak terlalu penting, kita dapat menggunakan `dropna()`.

Contoh:
```python
df.dropna(subset=["embarked"])
```

Jadi, `dropna()` dan `fillna()` dipilih berdasarkan kondisi dan tujuan analisis.

````python
# Contoh penggunaan dropna pada salinan data
df_contoh = df.copy()
df_contoh = df_contoh.dropna(subset=["embarked"])

print("Jumlah data setelah dropna:", len(df_contoh))
````

## 2. Data Duplikat

Duplikat adalah baris yang memiliki nilai sama pada seluruh kolom.

Duplikasi dapat membuat hasil analisis menjadi bias karena satu observasi seolah-olah muncul lebih dari sekali.

Gunakan:
- `duplicated().sum()` → menghitung jumlah duplikat.
- `drop_duplicates()` → menghapus duplikat dan menyisakan satu data unik.

````python
# Cek jumlah baris duplikat
print("Jumlah duplikat:", df_clean.duplicated().sum())

# Hapus duplikat
df_clean = df_clean.drop_duplicates()

print("Jumlah duplikat setelah dibersihkan:",
      df_clean.duplicated().sum())
````

## 3. Validasi dan Konversi Tipe Data

Tipe data harus sesuai dengan makna kolom.

Contohnya, `survived` hanya memiliki nilai `0` dan `1`. Secara makna, kolom tersebut merepresentasikan status **selamat atau tidak**, sehingga dapat direpresentasikan sebagai boolean.

Gunakan:
- `df.info()` untuk mengecek tipe data.
- `astype()` untuk konversi tipe yang jelas.
- `pd.to_numeric()` jika perlu mengubah data menjadi numerik.

````python
# Cek tipe data sebelum dikonversi
df_clean.info()
````

````python
# Kita labelin 0 = tidak selamat, 1 = selamat
df_clean["survived"] = df_clean["survived"].astype(bool)

# Cek hasil konversi
print(df_clean["survived"].dtype)
print(df_clean[["survived"]].head())
````

## 4. Scaling Numerik

Kita ambil dua fitur pada dataset:
- `age` dengan rentang sekitar 0–80;
- `fare` dengan rentang yang berbeda-beda.

Pada algoritma tertentu seperti K-Means atau SVM, perbedaan skala dapat memengaruhi hasil analisis.

**Scaling** digunakan agar fitur numerik berada pada skala yang lebih sebanding.

Metode:
- **MinMaxScaler** → mengubah nilai ke rentang `[0, 1]`.
- **StandardScaler** → menghasilkan data dengan mean ada di sekitar `0` dan standar deviasi `1`.

````python
# Normalisasi fare menjadi rentang 0–1
scaler_minmax = MinMaxScaler()

df_clean["fare_normalized"] = scaler_minmax.fit_transform(
    df_clean[["fare"]]
)

print(df_clean[["fare", "fare_normalized"]].head())
````

### StandardScaler

StandardScaler mengubah skala berdasarkan mean dan standar deviasi.

Hasilnya tidak harus berada pada rentang 0–1. Nilainya dapat berupa negatif maupun positif.

````python
scaler_standard = StandardScaler()

df_clean["fare_standardized"] = scaler_standard.fit_transform(
    df_clean[["fare"]]
)

print(df_clean[["fare", "fare_standardized"]].head())
````

## 5. Mengecek Hasil Preprocessing

Setelah cleaning dan preprocessing, kita perlu melakukan pengecekan kembali.

Yang penting bukan sekadar menjalankan fungsi, tetapi mengetahui **masalah apa yang ada dan mengapa suatu metode dipilih**.

````python
print("Ukuran data:", df_clean.shape)

print("\nTotal missing value:", df_clean.isnull().sum().sum())

print("Jumlah duplikat:", df_clean.duplicated().sum())

print("\nData setelah preprocessing:")
df_clean.head()
````

## 6. Menyimpan Data Bersih

Setelah proses selesai, data dapat disimpan agar bisa digunakan kembali.

CSV merupakan salah satu format yang sederhana dan umum digunakan.

````python
# Menyimpan dataset yang sudah dibersihkan
df_clean.to_csv("titanic_cleaned.csv", index=False)

print("Dataset berhasil disimpan sebagai titanic_cleaned.csv")
````
